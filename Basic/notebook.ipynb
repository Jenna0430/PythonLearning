Sets

Sets are unique, don't allow duplicates
Does not allow indexing because they are unordered

```
jobSkills = {'tableau', 'sql', 'python', 'statistics'}
```

```
jobSkills
```

```
jobSkills.add('looker')
jobSkills.add('sql')
jobSkills
```

Sets are very useful in converting a list of data into unique data that you can manipulate

```
skillList = ['python', 'sql', 'statistics', 'tableau', 'python', 'sql', 'statistics', 'tableau']

set(skillList)
```

```
tuple(range(5))
tuple(range(2,5))
tuple(range(1, 100, 2))
```

```
years_experience = {
    'Luke': 3,
    'Amanda': 4,
    'Kelly': 7,
    'Alex': 2,
}

for values in years_experience.values():
    print(values)

for key, value in years_experience.items():
    if value > 5:
      print(key, value)
```

```
years = 0
years_list = list(years_experience.items()) #returns a list of tuple
index = 0

while years < 5:
    key, years = years_list[index]
    print(key, years)
    index += 1
```
