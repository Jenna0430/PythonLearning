User defined functions


```python
base_salary = 100000
bonus_rate = 0.1

def calculate_salary(base, bonus):
    total = base * (1 + bonus)
    return total

calculate_salary(base_salary, bonus_rate)
calculate_salary(150000, 0.5)
```

```python
data_jobs = [
    {'job_title': 'Data Scientist', 'job_skills': "['python', 'sql', 'machine learning']", 'job_date': '2023-05-12'},
    {'job_title': 'Data Engineer', 'job_skills': "['AWS', 'python', 'tensorflow']", 'job_date': '2023-05-18'},
    {'job_title': 'Data Analyst', 'job_skills': "['python', 'sql', 'data analysis']", 'job_date': '2023-05-06'}
]
```

```python
from datatime import datetime
import ast

test_date = data_jobs[0]['job_date']

datetime.strptime(test_date, '%Y-%m-%d')
```

```python
for job in data_jobs:
    job['job_date'] = datetime.strptime(job['job_date'], '%Y-%m-%d')
    job['job_skills'] = ast.literal_eval(job['job_skills'])
```
