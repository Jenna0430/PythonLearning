Exercise: Skill investigation

```python
my_skills = ['python', 'sql', 'Excel']

job_roles = [
    {'role': 'Data Analyst', 'skills': ['python', 'sql', 'Excel']},
    {'role': 'Data Scientist', 'skills': ['python', 'R', 'Machine Learning', 'Deep Learning']},
    {'role': 'Machine Learning Engineer', 'skills': ['python', 'tensorflow', 'PyTorch', 'Sckrit-learn']},
    {'role': 'Data Engineer', 'skills': ['python', 'apache spark', 'hadoop', 'sql']},
    {'role': 'Business Intelligence Analyst', 'skills': ['python', 'sql', 'tableau', 'power bi', 'excel']},
    {'role': 'Quantitative Analyst', 'skills': ['python', 'R', 'Matlab', 'Statistics']},
    {'role': 'Operations Analyst', 'skills': ['python', 'data visualization', 'process improvement', 'sql']},
    {'role': 'Database Administrator', 'skills': ['excel', 'oracle', 'mysql', 'database management']},
    {'role': 'AI Engineer', 'skills': ['python', 'tensorflow', 'pytorch', 'computer vision']},
    {'role': 'Statistician', 'skills': ['python', 'sas', 'R', 'statistical modeling']},
]
```

```python
qualified_roles = []

for job in job_roles:
    qualified = True

    for skill in my_skills:
        if skill not in job['skills']:
            qualified = False
            break
    if qualified:
        qualified_roles.append(job['role'])

qualified_roles
```

```python
qualified_roles = []

for job in job_roles:
    if all(skill in job['skills'] for skill in my_skills):
        qualified_roles.append(job['role'])

qualified_roles
```

```python
qualified_roles = [job['role'] for job in job_roles if all(skill in job['skills'] for skill in my_skills)]

qualified_roles
```
