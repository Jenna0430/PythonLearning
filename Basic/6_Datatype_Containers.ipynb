DataType Containers

Lists : for a collection of ordered items

```python
jobSkills = ['sql', 'tableau', 'excel']
jobSkills
```

```python
jobSkills.append('python')
jobSkills
```

```python
jobSkills.remove('tableau')
jobSkills
```

```python
len(jobSkills)
```

```python
jobSkills.insert(2, 'tableau')
jobSkills
```

```python
jobSkills[0:2] #slicing 
```

```python
jobSkills[::2]
```

```python
jobSkills[-1:]
```

```python
skill1, skill2, skill3, skill4 = jobSkills
```

```python
f"{skill1}, {skill2}, {skill3}, {skill4}"
```

```python
skill1, *rest = jobSkills
f"{skill1},{rest}"
```

Mutable objects can be altered after creation like lists, dict, sets
While Immutable objects cannot be altered after creation like strings, tuples

Dictionary: stores data as key:value pairs, with the key being used to identify the value

```python
jobTypeSkills = {
    'database': 'pastgres',
    'analystTools': ['sap', 'jupyterLab'],
    'language' : ['R', 'sql', 'python']
}
jobTypeSkills
```

```python
jobTypeSkills['analystTools']
```

```python
jobTypeSkills.keys()
jobTypeSkills.values()
```

```python
jobTypeSkills['versionControl'] = 'git'
jobTypeSkills
```
