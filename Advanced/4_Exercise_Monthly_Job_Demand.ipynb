```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_India = df[df['job_country'] == 'India']
```

```python
df_India
```

```python
df_India['job_posted_month'] = df_India['job_posted_date'].dt.strftime('%B')
```

```python
df_India.sample(7)
```

```python
df_India_pivot = df_India.pivot_table(index='job_posted_month', columns='job_title_short', aggfunc='size')
df_India_pivot
```

```python
df_India_pivot.reset_index(inplace=True)
df_India_pivot
```

```python
df_India_pivot['month_no'] = pd.to_datetime(df_India_pivot['job_posted_month'], format='%B').dt.month
```

```python
df_India_pivot
```

```python
df_India_pivot.sort_values('month_no', inplace=True)
df_India_pivot
```

```python
df_India_pivot.set_index('job_posted_month', inplace=True)
df_India_pivot
```

```python
df_India_pivot.drop(columns=['month_no'], inplace=True)
```

```python
df_India_pivot.plot(kind='line')
```

```python
top_3 = df_India['job_title_short'].value_counts()[:3]
top_3
```

```python
top_3 = top_3.index.tolist()
top_3
```

```python
df_India_pivot[top_3].plot(kind='line')
```

Pandas Merge DataFrames

```python
software_jobs = pd.read_csv("https://lukeb.co/software_csv", index_col='job_posted_month')
```

```python
software_jobs
```

```python
merged_jobs = df_India_pivot.merge(software_jobs, on='job_posted_month')
```

```python
merged_jobs
```

```python
top_5 = merged_jobs.sum().sort_values(ascending=False)[:5].index.tolist()

top_5
```

```python
merged_jobs[top_5].plot(kind='line')
```
